```python
#Tree implemented via xgboost

#Dependencies

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
```

```python
df_diabetes = pd.read_csv("diabetes_012_health_indicators_BRFSS2015.csv")

target = "Diabetes_012"

df_diabetes_X = df_diabetes.drop(columns=[target])

df_diabetes_y = df_diabetes[target]

#Pre diabetics are treated as the same as diabetics for the purpose of classification.
df_diabetes_y[df_diabetes_y == 2] = 1


X_train, X_test, y_train, y_test = train_test_split(df_diabetes_X,df_diabetes_y, test_size = 0.2)

X_test, X_val, y_test, y_val = train_test_split(X_test, y_test, test_size=0.5)
```

```python
bst = XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.05, objective='binary:logistic')

bst.fit(X_train, y_train)

preds = bst.predict(X_test)
y_pred = (bst.predict_proba(X_test)[:, 1] >= 0.05).astype(int)
print(preds)

correct = 0
wrong = 0


for index, value in enumerate(y_pred):
    if value == y_test.iloc[index]:
        correct += 1
    else:
        wrong += 1

print(f"{(correct/(wrong+correct))}")
```

```python
from sklearn.metrics import classification_report
count = 0
print(classification_report(y_test, y_pred))
for value in y_pred:
    if value == 1.0:
        count +=1
print(count)
```
